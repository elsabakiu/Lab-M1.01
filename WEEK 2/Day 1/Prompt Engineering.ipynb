```python
from openai import OpenAI
import io
import os
from dotenv import load_dotenv
from pathlib import Path
import time

env_path = Path.cwd().parent / ".env"
load_dotenv(dotenv_path=env_path)

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

# Single-Call Helper Function
def run_prompt(
    prompt,
    model="gpt-4o-mini",
    temperature=0,
    max_tokens=300
):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "user", "content": prompt}
        ],
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content.strip()

# Multi-Run Helper Function
def run_multiple_times(
    prompt,
    runs=5,
    delay=1,
    **kwargs
):
    outputs = []

    for i in range(runs):
        result = run_prompt(prompt, **kwargs)
        outputs.append(result)

        print(f"Run {i + 1}/{runs} completed. Result: " + result)
        time.sleep(delay)

    return outputs


if __name__ == "__main__":
    prompt = "What is the capital of France?"
    answer = run_prompt(prompt)
    print(answer)

    test_prompt = "Respond with the word SUCCESS only."

    results = run_multiple_times(
        test_prompt,
        runs=3,
        temperature=0
    )

    print(results)
```

```python
# Initial simple prompt for sentiment analysis
sentiment_prompt_v1 = """

Classify the sentiment of the following customer message as positive, negative, or neutral.
Customer message: "I love the new design of your website, but the checkout process is really frustrating."
"""

# Test it once
result = run_prompt(sentiment_prompt_v1)
print("Sentiment Analysis Result:")
print(result)

# Initial simple prompt for product description
product_prompt_v1 = """
Write a short product description using the information below.

Product name: SmartFlow Water Bottle
Features: Keeps drinks cold for 24 hours, stainless steel body, leak-proof lid
Target audience: Busy professionals
"""
 
# Test it once
result = run_prompt(product_prompt_v1)
print("Product Description Result:")
print(result)


# Initial simple prompt for data extraction
extraction_prompt_v1 = """
Extract the key information from the customer feedback below.

Customer feedback:
"I love the design of the app, but it crashes every time I try to upload a file."

Identify:
- Main issue
- Overall sentiment
"""
 
# Test it once
result = run_prompt(extraction_prompt_v1)
print("Data Extraction Result:")
print(result)

```

```python
def append_to_file(filename, content):
    # Create file if it does not exist
    if not os.path.exists(filename):
        with open(filename, "w", encoding="utf-8") as file:
            pass

    # Append content
    with open(filename, "a", encoding="utf-8") as file:
        file.write(content + "\n")

result = run_multiple_times(sentiment_prompt_v1, 5)
append_to_file("prompt_results_5runs.txt", "\n".join(result))

result = run_multiple_times(product_prompt_v1, 5)
append_to_file("prompt_results_5runs.txt", "\n".join(result))

result = run_multiple_times(extraction_prompt_v1, 5)
append_to_file("prompt_results_5runs.txt", "\n".join(result))



```
